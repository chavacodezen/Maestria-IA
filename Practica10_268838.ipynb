Analisis de Ventas con GridSpec

**OBJETIVO**

Utilizar GridSpec para organizar subplots de diferentes tamaños en un análisis de ventas.

**INSTRUCCIONES**

1.   Genera un DataFrame de ventas con Mes, Ventas, y Gastos. Utiliza randint para generar números aleatorios enteros. Para ventas los valores deben estar entre 2000 y 5000. Para gastos, los valores debn estar entre 1000 y 4000.
2.   Usa GridSpec para organizar los subplots de la siguiente manera:
*   Primer gráfico (barra, ocupando dos columnas)
*   Segundo gráfico (línea, una columna)
*   Tercer gráfico (dispersión, dos filas).
3.   Agrega títulos y ajusta espacios.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Reproducibilidad
np.random.seed(2)

# 1. Genera un DataFrame de ventas con Mes, Ventas, y Gastos
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
df = pd.DataFrame({
    'Mes': meses,
    'Ventas': np.random.randint(2000, 5000, size=len(meses)),
    'Gastos': np.random.randint(1000, 4000, size=len(meses))
})

print(df, "\n")

# 2. Usa GridSpec para organizar los subplots
fig = plt.figure(figsize=(10, 6))
gs = gridspec.GridSpec(2, 3)

# Grafica Barras
ax1 = fig.add_subplot(gs[0, :2])
ax1.bar(df['Mes'], df['Ventas'], color='gray', edgecolor='black')
ax1.set_title('Ventas Mensuales')
ax1.set_ylabel('Ventas ($)')

# Grafica Linea
ax2 = fig.add_subplot(gs[0, 2])
ax2.plot(df['Mes'], df['Gastos'], marker='o', color='steelblue')
ax2.set_title('Gastos Mensuales')
ax2.set_ylabel('Gastos ($)')

# Grafica Dispersion
ax3 = fig.add_subplot(gs[1, :])
ax3.scatter(df['Ventas'], df['Gastos'], color='purple', s=80)
ax3.set_title('Relacion Ventas vs Gastos')
ax3.set_xlabel('Ventas ($)')
ax3.set_ylabel('Gastos ($)')

# 3. Títulos y espacios.
plt.tight_layout()
plt.show()
```
