Distribucion de Datos con plt.subplots()

**OBJETIVO**

Utilizar plt.subplots() para visualizar diferentes distribuciones generadas con NumPy y almacenadas en un DataFrame.

**INSTRUCCIONES**

1.   Genera un DataFrame con Pandas que contenga 1000 valores en cuatro columnas (Normal, Uniforme, Exponencial, Poisson).
2.   Crea una figura con plt.subplots(2, 2) para organizar los histogramas en una cuadrícula de 2x2.
3.   Grafica cada distribución en un subplot usando ax[i, j].hist().
4.   Agrega títulos a cada subplot y ajusta la disposición con tight_layout().

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generar valores reproducibles
np.random.seed(1)

# 1. Crear el DataFrame con 1000 valores
df = pd.DataFrame({
    'Normal': np.random.randn(1000),
    'Uniforme': np.random.uniform(-2, 2, 1000),
    'Exponencial': np.random.exponential(1, 1000),
    'Poisson': np.random.poisson(3, 1000)
})

# 2. Crear subplots 2x2
fig, ax = plt.subplots(2, 2, figsize=(10, 6))

# 3. Graficar cada histograma
ax[0, 0].hist(df['Normal'], bins=30, color='blue', alpha=0.7, edgecolor='black', linewidth=1)
ax[0, 0].set_title('Distribucion Normal')

ax[0, 1].hist(df['Uniforme'], bins=30, color='gray', alpha=0.7, edgecolor='black', linewidth=1)
ax[0, 1].set_title('Distribucion Uniforme')

ax[1, 0].hist(df['Exponencial'], bins=30, color='skyblue', alpha=0.7, edgecolor='black', linewidth=1)
ax[1, 0].set_title('Distribucion Exponencial')

ax[1, 1].hist(df['Poisson'], bins=30, color='red', alpha=0.7, edgecolor='black', linewidth=1)
ax[1, 1].set_title('Distribucion Poisson')

# 4. Ajustar la disposición
plt.tight_layout()
plt.show()
```
