## ✅ **Definición del problema**

El inventario es un componente crítico en la operación logística de una organización, ya que impacta la disponibilidad de materiales, la continuidad productiva y la eficiencia del almacén. Sin embargo, la complejidad aumenta cuando el inventario se distribuye entre múltiples ubicaciones y presenta diferencias entre cantidades disponibles, reservadas y en existencia.

El problema que se desea estudiar consiste en caracterizar el comportamiento del inventario a partir de snapshots diarios para comprender su estructura, identificar patrones de disponibilidad y detectar posibles concentraciones, desequilibrios o inventario potencialmente inmovilizado.

Este análisis es relevante porque permite obtener una visión descriptiva del estado del inventario y constituye la base para análisis posteriores de optimización logística y toma de decisiones.
