```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')

print("Primeras filas del dataset Iris:")
print(df.head())

print("\nInformacion general del Dataset:")
df.info()
```

a) Gráfico de barras:

*   Objetivo: Comparar el promedio de largo y ancho de los pétalos entre las tres especies de Iris.
*   Instrucciones: Utiliza un gráfico de barras para mostrar la media de las columnas petal_length y petal_width agrupadas por species.

```python
# Calcular los promedios por especie
promedios = df.groupby('species')[['petal_length', 'petal_width']].mean()
print(promedios)

# Posiciones en el eje X
especies = promedios.index
x = range(len(especies))
# Valores
petal_length_mean = promedios['petal_length']
petal_width_mean = promedios['petal_width']
# Ancho de las barras
width = 0.35

plt.figure()

# Barras
plt.bar(x, petal_length_mean, width, color='skyblue', label='Largo del Petalo')
plt.bar(
    [i + width for i in x],
    petal_width_mean,
    width,
    color='gray',
    label='Ancho del Petalo'
)

# Etiquetas
plt.xlabel('Especie')
plt.ylabel('Promedio')
plt.xticks([i + width / 2 for i in x], especies)
plt.legend()

plt.show()
```

b) Histograma:

*   Objetivo: Visualizar la distribución de los largos de los sépalos (sepal_length).
*   Instrucciones: Crea un histograma de la variable sepal_length y asegúrate de que tenga un número adecuado de bins. Coloca etiquetas en los ejes y un título.

```python
plt.figure()

# Histograma
plt.hist(
    df['sepal_length'],
    bins=20,              # barras
    edgecolor='gray',    # bordes
    color='skyblue'
)

# Etiquetas
plt.xlabel('Largo del Sepalo')
plt.ylabel('Frecuencia')
plt.title('Distribucion del largo del sepalo')

plt.show()
```
