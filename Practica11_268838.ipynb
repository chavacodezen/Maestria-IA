Seaborn y Subplots

**OBJETIVO**

Utilizar Seaborn junto con plt.subplots() para visualizar diferentes aspectos de un conjunto de datos en una sola figura.

**INSTRUCCIONES**

1.   Carga el dataset "penguins" de Seaborn. Utiliza df = sns.load_dataset("penguins")
2.   Usa plt.subplots() para crear una figura con 2 filas y 2 columnas.
3.   Genera los siguientes gráficos:
*   (0,0): sns.histplot(): Histograma de las longitudes de aletas.
*   (0,1): sns.boxplot(): Boxplot comparando la masa corporal por especie.
*   (1,0): sns.scatterplot(): Dispersión entre masa corporal y largo de aletas.
*   (1,1): sns.violinplot(): Violin plot de la masa corporal según la especie.
4.   Personaliza los gráficos con títulos y colores adecuados.
5.   Ajusta el diseño con plt.tight_layout().

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Cargar dataset
df = sns.load_dataset("penguins")

# 2. Crear figura y subplots (2x2)
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

# 3-4. Genera los siguientes gráficos:
# (0,0) Histograma de longitud de aletas
sns.histplot(
    data=df,
    x="flipper_length_mm",
    bins=20,
    color="skyblue",
    ax=ax[0, 0]
)
ax[0, 0].set_title("Distribucion de Longitud de Aletas")
ax[0, 0].set_xlabel("Longitud de la aleta (mm)")
ax[0, 0].set_ylabel("Frecuencia")

# (0,1) Boxplot de masa corporal por especie
sns.boxplot(
    data=df,
    x="species",
    y="body_mass_g",
    hue="species",
    palette="Pastel1",
    legend=False,
    ax=ax[0, 1]
)
ax[0, 1].set_title("Masa Corporal por Especie")
ax[0, 1].set_xlabel("Especie")
ax[0, 1].set_ylabel("Masa corporal (g)")

# (1,0) Scatterplot masa corporal vs longitud de aletas
sns.scatterplot(
    data=df,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    palette="Pastel1",
    ax=ax[1, 0]
)
ax[1, 0].set_title("Masa Corporal vs Longitud de Aletas")
ax[1, 0].set_xlabel("Longitud de la aleta (mm)")
ax[1, 0].set_ylabel("Masa corporal (g)")

# (1,1) Violin plot de masa corporal por especie
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    hue="species",
    palette="Pastel1",
    legend=False,
    ax=ax[1, 1]
)
ax[1, 1].set_title("Distribucion de Masa Corporal por Especie")
ax[1, 1].set_xlabel("Especie")
ax[1, 1].set_ylabel("Masa corporal (g)")

# 5. Ajustar diseño
plt.tight_layout()
plt.show()
```
